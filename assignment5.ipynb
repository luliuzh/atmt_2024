## Visualization

Visualize the resulting BLEU scores in an appropriate plot and discuss them in one to two paragraphs of your PDF report. 

What happens to the brevity penalty calculated by SacreBLEU as you increase the beam size? 

```python
!pip install matplotlib
```

```python
import matplotlib.pyplot as plt

beam_size = [5, 10, 15, 20, 25]
blue = [20.0, 22.2, 22.1, 21.8, 21.4]
bp = [1.000, 1.000, 0.949,0.864, 0.790]

# Create figure and axis objects
fig, ax1 = plt.subplots()
ax1.set_xticks(beam_size)

# Plot blue values on the left y-axis
ax1.plot(beam_size, blue, 'b-', label='Blue', marker='o')
ax1.set_xlabel('Beam Size')
ax1.set_ylabel('Blue', color='b')
ax1.tick_params(axis='y', labelcolor='b')

# Create a second y-axis to plot bp values
ax2 = ax1.twinx()
ax2.plot(beam_size, bp, 'r-', label='BP', marker='o')
ax2.set_ylabel('BP', color='r')
ax2.tick_params(axis='y', labelcolor='r')

# Title and showing the plot
plt.title('Blue and BP Scores vs. Beam Size')
plt.show()
```
